```python
import pandas as pd

# 1. ĐỌC DỮ LIỆU ĐÃ MERGE CHUẨN
df = pd.read_csv("C:\\Users\\Viet Hung\\Downloads\\gois1-1109\\data_clean\\master_dataset.csv")
df['Date'] = pd.to_datetime(df['Date'])

print("=== TỔNG QUAN DỮ LIỆU SAU KHI FIX ===")
print(df.info())

# 2. TÌM NGOẠI LỆ & BẤT NGỜ (So sánh đỉnh điểm cú sốc 2022-2023 với hiện tại 2024-2026)
df_shock = df[(df['Date'].dt.year >= 2022) & (df['Date'].dt.year <= 2023)].groupby('Country').mean(numeric_only=True)
df_latest = df[df['Date'].dt.year >= 2024].groupby('Country').mean(numeric_only=True)

# Tổng hợp các chỉ số thay đổi để tìm insight
insights = pd.DataFrame({
    'Muc_Giam_Lam_Phat': df_shock['Inflation'] - df_latest['Inflation'],
    'Thiet_Hai_Tin_Dung': df_shock['Credit_Growth'] - df_latest['Credit_Growth'],
    'That_Nghiep_Hien_Tai': df_latest['Unemployment']
})

print("\n=== BẢNG TÌM KIẾM INSIGHT CHO LUẬN VĂN ===")
print(insights.sort_values(by='Muc_Giam_Lam_Phat', ascending=False))
```
